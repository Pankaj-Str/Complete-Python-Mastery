```python
# function with parameter / arg and return type

def add(number1,number2):
    result = number1 + number2
    return result
```

```python
# function with default parameters 
def bill(price,gst=18):
    gst_amount = price * gst / 100
    total_price = price + gst_amount
    print("\n-------------bill--------------\n")
    print(f"Price = {price} ₹")
    print(f"GST = {gst} %")
    print(f"GST Amount = {gst_amount} ₹")
    print(f"MRP include GST = {total_price}₹")
```

```python
bill(1200)
bill(1200,9)
```

```python
# function with keyword and arguments 
# we can pass arguments using parameter name 

def student(name,age,course):
    print("Name : ",name)
    print("Age : ",age)
    print("Course :",course)


#student("Python",23,"joy")
student(
    course="Python",
    age=25,
    name="Joy"
)
```
