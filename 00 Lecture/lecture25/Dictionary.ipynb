```python
data = {
    "Name":"Nishant",
    "Age":23,
    "Height":4.5
}

print(data)

# checking data type
print(type(data))
```

```python
# Accessing Values
print(data["Name"])
print(data["Age"])
```

```python
# Modifying 
data["Age"] = 12
print(data)

# Adding Items
data["City"] = "Mumbai"
print(data)
```

```python
# Checking Key Existence
if "City" in data:
    print("Marks Key is Found...")
else:
    print("Marks Key is not Found...")
```

```python
# keys(): Returns a list of all keys in the dictionary.
print(data.keys())
# values(): Returns a list of all values in the dictionary.
print(data.values())
# items(): Returns a list of key-value pairs as tuples.
print(data.items())
# get(key, default): Returns the value for a key if it exists, else returns a default value.
print(data.get("Name"))
```

```python
# pop(key, default): Removes and returns the value associated with the key.
data.pop("City")
print(data)
```

```python
# using loop 
for i in data:
    # print(i,' = ',data[i])
    print(i,' = ',data.get(i))
```
