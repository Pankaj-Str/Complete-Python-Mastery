```python
def book_ticket(passenger_name, train_name):

   print("Passenger Name :", passenger_name)
   print("Train Name :", train_name)
   print("Reservation Status : Confirmed")
```

```python
book_ticket(
   "Rahul Sharma",
   "Rajdhani Express"
)
```

```python
def calculate_fare(ticket_price, passengers):

   total_amount = ticket_price * passengers

   return total_amount
```

```python
final_fare = calculate_fare(2450, 2)

print("Final Fare :", final_fare)
```

```python
def verify_passenger(age):

   if age >= 18:

       return "Eligible for Reservation"

   else:

       return "Not Eligible"
```

```python
result = verify_passenger(25)

print(result)
```

```python
def generate_ticket(
   passenger_name,
   train_name,
   seat_number
):

   print("===================================")
   print(" Railway Reservation Ticket ")
   print("===================================")

   print("Passenger Name :", passenger_name)
   print("Train Name :", train_name)
   print("Seat Number :", seat_number)
```

```python
generate_ticket(
   "Priya Singh",
   "Shatabdi Express",
   "B2-21"
)
```

```python
def reservation_summary(
   passenger_name,
   source_station,
   destination_station
):

   print("Passenger :", passenger_name)
   print("Source :", source_station)
   print("Destination :", destination_station)
```

```python
reservation_summary(
   "Amit Verma",
   "Mumbai",
   "Delhi"
)
```
