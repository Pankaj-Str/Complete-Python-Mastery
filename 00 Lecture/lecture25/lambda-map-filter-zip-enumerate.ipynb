```python
# normal function 
def square(num):
    return num * num

# call function 
result = square(5)
print(result)
```

```python
# lambda function 
square = lambda num : num * num
result = square(5)
print(result)
```

```python
add = lambda a , b : a + b
print(add(10,20)) # call lambda function
```

```python
# find the maximum number 

max = lambda a , b : a if a > b else b
print(max(8,34))
```

```python
# lamdbda with string 

get_length = lambda name : len(name)
print(get_length("python"))
```

```python
# lambda function with list

numbers = [21,22,23,45,56,78]
square = lambda x : x * x
result = [square(x) for x in numbers]
print(result)
```

```python
# map()
# map(function,iterable)
# iterable - > list , tuple
# without map
numbers = [21,22,23,24,25]
result = []
for i in numbers:
    result.append(i * i)

print(result)
```

```python
# with map 
numbers = [21,22,23,24,25]
def square(num):
    return  num * num

result = map(square,numbers)

print(list(result)) # convert into list
```

```python
# Using Map and lambda function
numbers = [21,22,23,24,25]

result = map(lambda x : x * x ,numbers)

print(list(result)) # convert into list
```

```python
data = ["Java","Python","C","Jsp"]

result = map(lambda data: data.upper(),data)

print(list(result))
```
