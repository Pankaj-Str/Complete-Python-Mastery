```python
# empty set 
data = {}
print(data)

# check type
print(type(data))

data = set()
print(data)
# check type
print(type(data))
```

```python
marks = [21,45,45,78,96,96,45,45,12,12,45,45,12]
print(marks)
set_marks = set(marks)
print(set_marks)
```

```python
data = {23,23,23,23,24,25,26,26,25}
print(data)
```

```python
# Adding Elements to a Set
data.add(100)
print(data)
```

```python
# add multiple data
new_data = [34,35,36,100,23]
data.update(new_data)
print(data)
```

```python
# remove()	Removes item	Yes (raises error)
data.remove(24)
print(data)
```

```python
# discard()	Removes item (safer)	No
data.discard(23)
print(data)
```

```python
# pop()	Removes and returns one random item	No (works on empty too)
data.pop()
print(data)
```

```python
# clear()	Removes all items	No
data.clear()
print(data)
```
