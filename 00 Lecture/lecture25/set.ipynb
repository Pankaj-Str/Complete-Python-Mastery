```python
# Python Set Methods

# 1. Creating a Set
numbers = {10, 20, 30, 40, 50}
print(numbers)

# 2. add() - Add Single Element
numbers.add(60)
print(numbers)

# 3. update() - Add Multiple Elements
numbers.update([70, 80, 90])
print(numbers)

# 4. remove() - Remove Element
numbers.remove(20)
print(numbers)

# 5. discard() - Remove Element Without Error
numbers.discard(100)
print(numbers)

# 6. pop() - Remove Random Element
numbers.pop()
print(numbers)

# 7. copy() - Copy Set
new_set = numbers.copy()
print(new_set)

# 8. clear() - Remove All Elements
temp = {1, 2, 3}
temp.clear()
print(temp)

# 9. union() - Combine Two Sets
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a.union(b))
print(a | b)

# 10. intersection() - Common Elements
print(a.intersection(b))
print(a & b)

# 11. difference() - Elements in First Set Only
print(a.difference(b))
print(a - b)

# 12. symmetric_difference() - Unique Elements
print(a.symmetric_difference(b))
print(a ^ b)

# 13. intersection_update()
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

a.intersection_update(b)
print(a)

# 14. difference_update()
a = {1, 2, 3, 4}
b = {3, 4}

a.difference_update(b)
print(a)

# 15. symmetric_difference_update()
a = {1, 2, 3}
b = {3, 4, 5}

a.symmetric_difference_update(b)
print(a)

# 16. issubset() - Check Subset
a = {1, 2}
b = {1, 2, 3, 4}

print(a.issubset(b))

# 17. issuperset() - Check Superset
print(b.issuperset(a))

# 18. isdisjoint() - Check No Common Elements
a = {1, 2, 3}
b = {4, 5, 6}

print(a.isdisjoint(b))

# 19. len() - Find Set Length
numbers = {10, 20, 30, 40}
print(len(numbers))

# 20. in - Membership Operator
print(20 in numbers)
print(100 not in numbers)

# 21. min() and max()
print(min(numbers))
print(max(numbers))

# 22. sum() - Calculate Total
print(sum(numbers))

# 23. sorted() - Sort Set Elements
print(sorted(numbers))

# 24. Loop Through Set
for value in numbers:
    print(value)

# 25. Remove Duplicate Values
data = [10, 20, 10, 30, 20, 40]
unique_data = set(data)

print(unique_data)
```

```python
# Python Set Operations

A = {10, 20, 30, 40, 50}
B = {30, 40, 50, 60, 70}

print("Set A:", A)
print("Set B:", B)


# 1. Union (A ∪ B)
# Combines all unique elements from both sets

print("Union:", A | B)
print("Union:", A.union(B))


# 2. Intersection (A ∩ B)
# Returns common elements

print("Intersection:", A & B)
print("Intersection:", A.intersection(B))


# 3. Difference (A - B)
# Elements present in A but not in B

print("Difference A-B:", A - B)
print("Difference B-A:", B - A)


# 4. Symmetric Difference (A △ B)
# Returns elements that are not common

print("Symmetric Difference:", A ^ B)
print("Symmetric Difference:", A.symmetric_difference(B))


# 5. Subset
# Checks whether all elements of A exist in B

X = {10, 20}
Y = {10, 20, 30, 40}

print("Subset:", X.issubset(Y))


# 6. Superset
# Checks whether Y contains all elements of X

print("Superset:", Y.issuperset(X))


# 7. Disjoint
# Checks whether two sets have no common elements

P = {1, 2, 3}
Q = {4, 5, 6}

print("Disjoint:", P.isdisjoint(Q))


# 8. Union Assignment
X = {1, 2, 3}
Y = {3, 4, 5}

X |= Y
print("Union Assignment:", X)


# 9. Intersection Assignment
X = {1, 2, 3, 4}
Y = {3, 4, 5, 6}

X &= Y
print("Intersection Assignment:", X)


# 10. Difference Assignment
X = {1, 2, 3, 4}
Y = {3, 4}

X -= Y
print("Difference Assignment:", X)


# 11. Symmetric Difference Assignment
X = {1, 2, 3}
Y = {3, 4, 5}

X ^= Y
print("Symmetric Difference Assignment:", X)
```
