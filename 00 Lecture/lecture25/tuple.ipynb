```python
data = (34)
print(data)
# check type
print(type(data))
data = (34,)
print(type(data))
data1 = 56,
print(type(data1))

data2 = 56,89,90,67
print(type(data2))
data2 = (56,89,90,67)
print(type(data2))
```

```python
data2 = (56,89,90,67)
print(data2)
print(data2[1])
# data2[1] = 100 -> 'tuple' object does not support item assignment
print(data2)
```

```python
# Concatenation
# Concatenating tuples
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
result = tuple1 + tuple2
print(result)  # Output: (1, 2, 3, 4, 5, 6)
```

```python
print(tuple1 * 3)
```

```python
# count()
data5 = (1, 2, 3, 1, 2, 3, 1, 2, 3)
print(data5.count(2))

# index()
print(data5.index(3))
```

```python
# in and not in 
print(1 in data5)

print(3 not in data5)
```

```python
for i in data5:
    print(i)
```
