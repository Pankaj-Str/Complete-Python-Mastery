```python
data = "Codes with Python"
# 1. upper() converts the string to uppercase
print(data.upper())
# 2. lower() converts the string to lowercase
print(data.lower())
# 3. partition() returns a tuple
print(data.partition(" "))
# 4. replace() replaces substring inside
data = data.replace("Python","Pankaj")
print(data)
# 5. find() returns the index of first occurrence of substring
print(data.find("w"))
# 6. rstrip() removes trailing characters
data = "cwpc.in   "
print(data,4545)
print(data.rstrip(),465456)

# 7. split() splits string from left
data = "q w w e r t y y u u o p"
print(data.split(" "))
# 8. startswith() checks if string starts with the specified string
data = "codeswithpankaj"
print(data.startswith("W"))
# 9. isnumeric() checks numeric characters
data = "34"
print(data.isnumeric())
# 10. index() returns index of substring
data = "codeswithpankaj"
print(data.index("p"))			
```

```python
# format()

name = "joy"
age = 23
height = 4.5

print("My name is ",name," my age ",age,"and height ",height)

# using format()
print("My name is {0} my age {1} and height {2}".format(name,age,height))

# using f
print(f"My name is {name} my age {age} and height {height}")
```

```python
# Python String Methods

text = "  Welcome to Python Programming  "

# 1. upper() - Convert to Uppercase
print(text.upper())

# 2. lower() - Convert to Lowercase
print(text.lower())

# 3. title() - Convert First Letter of Each Word to Uppercase
print(text.title())

# 4. capitalize() - Capitalize First Character
print(text.capitalize())

# 5. swapcase() - Reverse Uppercase and Lowercase
print(text.swapcase())

# 6. strip() - Remove Spaces from Both Sides
print(text.strip())

# 7. lstrip() - Remove Left Spaces
print(text.lstrip())

# 8. rstrip() - Remove Right Spaces
print(text.rstrip())

# 9. replace() - Replace a String
print(text.replace("Python", "Java"))

# 10. split() - Convert String into List
data = "Python Java C++"
print(data.split())

# 11. join() - Join List into String
languages = ["Python", "Java", "C++"]
print(" ".join(languages))

# 12. find() - Find Position of Substring
print(text.find("Python"))

# 13. index() - Find Index of Substring
print(text.index("Python"))

# 14. count() - Count Occurrences
message = "Python is easy. Python is powerful."
print(message.count("Python"))

# 15. startswith() - Check Starting Characters
print(text.startswith("  Welcome"))

# 16. endswith() - Check Ending Characters
print(text.endswith("  "))

# 17. isalpha() - Check Only Alphabets
name = "Pankaj"
print(name.isalpha())

# 18. isdigit() - Check Only Digits
number = "12345"
print(number.isdigit())

# 19. isalnum() - Check Alphabets and Numbers
value = "Python123"
print(value.isalnum())

# 20. isspace() - Check Only Spaces
space = "   "
print(space.isspace())

# 21. islower() - Check Lowercase
print("python".islower())

# 22. isupper() - Check Uppercase
print("PYTHON".isupper())

# 23. istitle() - Check Title Case
print("Python Programming".istitle())

# 24. center() - Align String in Center
print("Python".center(20, "*"))

# 25. ljust() - Left Align
print("Python".ljust(15, "-"))

# 26. rjust() - Right Align
print("Python".rjust(15, "-"))

# 27. zfill() - Add Zeros at Beginning
num = "123"
print(num.zfill(6))

# 28. partition() - Split into Three Parts
sentence = "Python is easy"
print(sentence.partition("is"))

# 29. rfind() - Find Last Occurrence
message = "Python Java Python"
print(message.rfind("Python"))

# 30. rindex() - Find Last Index
print(message.rindex("Python"))

# 31. expandtabs() - Replace Tabs with Spaces
tab_text = "Python\tJava\tC++"
print(tab_text.expandtabs(4))

# 32. removeprefix() - Remove Prefix
url = "https://codeswithpankaj.com"
print(url.removeprefix("https://"))

# 33. removesuffix() - Remove Suffix
filename = "lecture.py"
print(filename.removesuffix(".py"))

# 34. splitlines() - Split String by Lines
lines = "Python\nJava\nC++"
print(lines.splitlines())

# 35. format() - Format String
name = "Pankaj"
course = "Python"

print("My name is {} and I teach {}.".format(name, course))

# 36. format_map() - Format Using Dictionary
student = {"name": "Rahul", "course": "Python"}
print("Name: {name}, Course: {course}".format_map(student))

# 37. encode() - Convert String into Bytes
text = "Python"
print(text.encode())

# 38. casefold() - Case-Insensitive Conversion
print("PYTHON".casefold())

# 39. isdecimal() - Check Decimal Characters
print("12345".isdecimal())

# 40. isnumeric() - Check Numeric Characters
print("12345".isnumeric())

# 41. isidentifier() - Check Valid Variable Name
print("student_name".isidentifier())

# 42. isprintable() - Check Printable Characters
print("Python 123".isprintable())

# 43. isascii() - Check ASCII Characters
print("Python".isascii())
```
