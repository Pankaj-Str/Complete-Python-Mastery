```python
# function with parameter / Arg
def add(number1,number2):
    result = number1 + number2
    print("This is Result = ",result)    
```

```python
# bill
def bill(price,gst):
    gst_amount = price * gst / 100
    total_price = price + gst_amount
    print("\n-------------bill--------------\n")
    print(f"Price = {price} ₹")
    print(f"GST = {gst} %")
    print(f"GST Amount = {gst_amount} ₹")
    print(f"MRP include GST = {total_price}₹")
```

```python
bill(1200,18)
bill(1800,2)
bill(8500,9) 
```
