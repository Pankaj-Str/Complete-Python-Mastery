```python
# import date and time 
import datetime
```

```python
# get current date and time 
current = datetime.datetime.now()
print(current)

# get current date 
today = datetime.date.today()
print(today)

# get current time 
current_time = datetime.datetime.now().time()
print(current_time)
```

```python
from datetime import date
# create a custom date 
my_date = date(2026,11,3)

print(my_date)
print("Year = ",my_date.year)
print("Month = ",my_date.month)
print("Day = ",my_date.day)
```

```python
from datetime import time

my_time = time(10,30,45)
print(my_time)
print("Hour : ",my_time.hour)
print("Minute : ",my_time.minute)
print("Second : ",my_time.second)
```

```python
# format date and time using strftime()
current = datetime.datetime.now()

print(current)
print(current.strftime("%d-%m-%y"))
print(current.strftime("%y/%m/%d"))
print(current.strftime("%H:%M:%S"))
print(current.strftime("%I:%M:%p"))

# p = AM/PM
# I = 12 hour format
# H = 24 hour format
# A full weekday
print(current.strftime("%a-%m-%y"))
print(current.strftime("%A-%m-%y"))
# %B Month name %b half month name
print(current.strftime("%y/%b/%d"))
print(current.strftime("%y/%B/%d"))
```

```python
date_str = "03-11-2026"

# datatype 
print(type(date_str))

current = datetime.datetime.now()

print(type(current))

# convert string into date using strptime()

data_date = datetime.datetime.strptime(date_str,"%d-%m-%Y")

print(data_date)
print(type(data_date))
```
