```python
class employee1:
    def __init__(self):
         print("CWPC Infotech PVT. LTD.")
    def employee_salary1(self,salary):
        print("Employee Salary : ",salary)

    def employee_name1(self,name):
            print("Employee name : ",name) 
```

```python
emp = employee1()
emp.employee_salary1(45000)
emp.employee_name1("joy")
```

```python
class employee2:
    def __init__(self):
         print("CWPC Infotech PVT. LTD.")
    def tds(self,tax):
         self.tax = tax
         print("TDS = ",tax)
    def employee_salary2(self,salary):
        print("Employee Salary : ",salary-self.tax)
    def employee_name2(self,name):
            print("Employee name : ",name) 
```

```python
emp = employee2()
emp.tds(100)
emp.employee_salary2(45000)
emp.employee_name2("joy")
```
