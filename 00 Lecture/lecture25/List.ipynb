```python
# Lists are one of the most versatile and commonly used data structures in Python. They are ordered collections of items, which can be of different types, including integers, strings, and even other lists.


data = [21,22,23,2.4,25]
print(data)
data = ["java","C++","JSP","Python","Ruby",True]
print(data)
data = [100,200,400,['a','b','c']]
print(data)
data = []
print(data)
# check type 
print(type(data))
```

```python
data = [21,22,23,24,25,26,27,28]
print(data[5])
print(data[2:6])
print(data[4:])
print(data[:4])
print(data[:])
# update data 
data[4] = 100
print(data)
```

```python
data = [21,22,23,24,25]
# List Methods
# append()
data.append(500)
print(data)
# extend()
data.extend([100,200,300])
print(data)
# insert()
data.insert(2,99)
print(data)
```

```python
data = [21,22, 99,99,99, 23, 24, 25, 500, 100, 200, 300]
# remove()
data.remove(99)
print(data)
```

```python
data = [21,22,99,99,99, 23, 24, 25, 500, 100, 200, 300]
# pop()
data.pop(4)
print(data)
```

```python
data = [21, 22, 99, 99, 23, 24, 25, 500, 100, 200, 300]
# index()
print(data.index(24))
# count()
print(data.count(99))
# sort()
data.sort()
print(data)
# reverse()
data.reverse()
print(data)
# copy()
data1 = data.copy()
print(data1)
# clear()
data.clear()
print(data)
```

```python
data = [21, 22, 99, 99, 23, 24, 25, 500, 100, 200, 300]
print(len(data))
print(sum(data))
print(min(data))
print(max(data))
```
