```python
# process data using filter() and map()
ticket_fares = [2500,5600,7820,600,800,1200]

valid_fares = list(filter(lambda fare : fare > 1500,ticket_fares))
print(valid_fares)
```

```python
# Apply fare update map 
final_fares = list(map(lambda fare : fare + 100 ,valid_fares))
print(final_fares)
```

```python
# build final module functional 
passengers = ["Amit","Joy","Nishant","Rahul"]
ticket_fares = [2500,5600,7820,600]
seats = ["B1-21","B1-22","B1-23","B1-24"]

# using zip functions 
reservation_data = list(zip(passengers,seats,ticket_fares))

# using for loop to print all list 

for index , data in enumerate(reservation_data):
    passengers , seats , ticket_fares = data
    final_fare = (lambda x : x - 100) (ticket_fares)
    print("Reservation ID : ", index + 1)
    print("Passengers : ",passengers)
    print("Seat : ",seats)
    print("Final Fare : ",final_fare)
    print("--------------------------------")
    
    
```
