## 1. Welcome Message

```python
def welcome_message():
    print("===================================")
    print("   Indian Railway Reservation")
    print("===================================")
    print("Welcome! Book your train ticket.")

welcome_message()
```

## 2. Booking a Ticket – Function with Parameters

```python
def book_ticket(passenger_name, train_name):
    print("Passenger Name:", passenger_name)
    print("Train Name:", train_name)
    print("Booking Status: Confirmed")

book_ticket("Rahul Sharma", "Rajdhani Express")
book_ticket("Priya Singh", "Shatabdi Express")
```

## 3. Calculate Ticket Fare – Function with Return

```python
def calculate_fare(ticket_price, number_of_passengers):
    total_amount = ticket_price * number_of_passengers
    return total_amount

fare1 = calculate_fare(2450, 2)
fare2 = calculate_fare(1500, 3)

print("Rahul's Total Fare: Rs.", fare1)
print("Priya's Total Fare: Rs.", fare2)
```

## 4. Calculate Fare with Discount

```python
def calculate_final_fare(ticket_price, passengers, discount):
    total = ticket_price * passengers
    final_amount = total - discount
    return final_amount

amount = calculate_final_fare(2000, 2, 500)
print("Final Amount: Rs.", amount)
```

## 5. Check Passenger Age – if-else

```python
def verify_passenger(age):
    if age >= 18:
        return "Adult passenger"
    else:
        return "Minor passenger"

print(verify_passenger(25))
print(verify_passenger(15))
```

## 6. Generate a Railway Ticket – Multiple Parameters

```python
def generate_ticket(passenger_name, train_name, seat_number):
    print("===================================")
    print("       RAILWAY RESERVATION TICKET")
    print("===================================")
    print("Passenger Name:", passenger_name)
    print("Train Name:", train_name)
    print("Seat Number:", seat_number)
    print("Status: Confirmed")
    print("===================================")

generate_ticket("Priya Singh", "Shatabdi Express", "B2-21")
```

## 7. Display Journey Details

```python
def journey_details(passenger_name, source, destination):
    print("Passenger:", passenger_name)
    print("From:", source)
    print("To:", destination)

journey_details("Amit Verma", "Mumbai", "Delhi")
journey_details("Neha Patel", "Pune", "Ahmedabad")
```

## 8. Default Parameter – Train Class

```python
def show_booking(passenger_name, train_class="Sleeper"):
    print("Passenger:", passenger_name)
    print("Train Class:", train_class)

show_booking("Neha")
show_booking("Amit", "AC 3 Tier")
```

## 9. Calculate Fare for Different Train Classes

```python
def get_ticket_price(train_class):
    if train_class == "Sleeper":
        return 500
    elif train_class == "AC 3 Tier":
        return 1200
    elif train_class == "AC 2 Tier":
        return 1800
    else:
        return 0

print("Sleeper Fare: Rs.", get_ticket_price("Sleeper"))
print("AC 3 Tier Fare: Rs.", get_ticket_price("AC 3 Tier"))
print("AC 2 Tier Fare: Rs.", get_ticket_price("AC 2 Tier"))
```

## 10. Complete Railway Booking Mini Project

```python
def welcome_message():
    print("===================================")
    print("     INDIAN RAILWAY BOOKING")
    print("===================================")


def calculate_fare(ticket_price, passengers):
    return ticket_price * passengers


def display_ticket(name, train, source, destination, seat, total_fare):
    print("\n========== BOOKING TICKET ==========")
    print("Passenger Name:", name)
    print("Train Name:", train)
    print("From:", source)
    print("To:", destination)
    print("Seat Number:", seat)
    print("Total Fare: Rs.", total_fare)
    print("Booking Status: Demo Confirmed")
    print("====================================")


def final_booking_system():
    passenger_name = "Rahul Sharma"
    train_name = "Rajdhani Express"
    source_station = "Mumbai"
    destination_station = "Delhi"
    seat_number = "A1-12"

    ticket_price = 2450
    number_of_passengers = 2

    welcome_message()
    total_fare = calculate_fare(ticket_price, number_of_passengers)

    display_ticket(
        passenger_name,
        train_name,
        source_station,
        destination_station,
        seat_number,
        total_fare
    )


final_booking_system()
```

## 11. Booking Multiple Passengers

```python
def print_passenger(name, age, train):
    print("Name:", name, "| Age:", age, "| Train:", train)

passengers = [
    ("Rahul", 25, "Rajdhani Express"),
    ("Priya", 22, "Shatabdi Express"),
    ("Amit", 31, "Duronto Express")
]

for passenger in passengers:
    print_passenger(passenger[0], passenger[1], passenger[2])
```
