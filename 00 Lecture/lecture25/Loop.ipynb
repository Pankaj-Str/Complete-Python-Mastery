```python
#loop

x = 1
while x <= 10:
    print(x)
    x = x + 1
```

```python
# for loop

# range(end_point)

for i in range(10):
    print(i)
```

```python
# range(start,end)

for i in range(10,20):
    print(i)
```

```python
# range(start,end,steps)
for i in range(11,30,3):
    print(i)
```

```python
# end=''
for i in range(1,20):
    print(i,end=' | ')
```

```python
for i in range(10,0,-1):
    print(i)
```

```python
for i in range(-10,-0):
    print('Remaining Seats: ',-i)
```

```python
for i in range(1,11):
    print(i*2)
```

```python
total_seats = 5

while total_seats > 0:
    print("Seat Booked Successfully")
    print("Remaining Seats:", total_seats - 1)
    total_seats -= 1

print("All Seats Are Filled")
```
