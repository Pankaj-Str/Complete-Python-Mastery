```python
# empty set using set function set()
my_set = set()
print(my_set)
```

```python
# data list 
data = [21,22,21,21,23,23,23,24,24,23,24,25,26]
print(data) # print list
my_set = set(data)
print(my_set) # print set 
```

```python
data_set = {1,1,1,2,3,4,4,4,4,5,6,6,6,7,7,7,9}
print(data_set)
```

```python
# add new data into set 
# using method add() and update()
data_set.add(100)
print(data_set)

# add multiple item into set 
data_set.update([20,30,20,20])
print(data_set)
```

```python
# remove()
data_set.remove(100)
print(data_set)
```

```python
# discart()
data_set.discard(100)
print(data_set)
```

```python
# pop()
data_set.pop()
print(data_set)
```

```python
# clear()
data_set.clear()
print(data_set)
```
