```python
data = {
    "Name":"Prachi",
    "Age":19,
    "height":4.9
}
```

```python
print(data)
```

```python
# access data from key
print(data["Name"])
print(data["Age"])
```

```python
# update data 
data["Age"] = 18
print(data)
```

```python
# add new item into dictionary
data["City"] = "Mumbai"
print(data)
```

```python
# keys(): Returns a list of all keys in the dictionary.
print(data.keys())
```

```python
# values(): Returns a list of all values in the dictionary.
print(data.values())
```

```python
# items(): Returns a list of key-value pairs as tuples.
print(data.items())
```

```python
# get(key, default): Returns the value for a key if it exists, else returns a default value
print(data.get("Name"))
```

```python
# pop(key, default): Removes and returns the value associated with the key.
data.pop("City")
print(data)
```

```python
# using loop for print dictionary
for i in data:
    print(i,' = ',data[i])
```
