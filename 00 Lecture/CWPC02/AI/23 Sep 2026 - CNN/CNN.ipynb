```python
from PIL import Image ,ImageDraw , ImageFont
```

```python
# create image
img = Image.new("RGB",(1200,600),"white")
draw = ImageDraw.Draw(img)
```

```python
# install fonts
!apt-get -qq install fonts-dejavu
```

```python
# Font
font = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf",35)
```

```python
# CNN Flow
steps = [
    "IMAGE",
    "CON",
    "FEA",
    "POOL",
    "PRED"
]

x = 50

for i , step in enumerate(steps):
  # Box
  draw.rounded_rectangle(
      (x,230,x+190,330),
      radius=15,
      outline="black",
      width=3
  )
  # text
  draw.text(
      (x + 25 ,260),
      step,
      fill="black",
      font=font
  )
  # Arrow
  if i < len(steps) - 1:
    draw.text(
        (x + 200,255),
        "→",
        fill="black",
        font=font
    )

  x += 230

  # save image
img.save("CNN_01.png")
print("image save successfully")
```
