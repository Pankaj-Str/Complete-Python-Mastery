```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# Traning Data
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

# answer
y = np.array([0,0,0,1])
```

```python
# create a ANN
model = Sequential(
    [
        Dense(4, activation='relu', input_shape=(2,)),
        Dense(1, activation='sigmoid')
    ]
)
```

```python
# Prepare model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
# train ANN
model.fit(X,y,epochs=100,verbose=0)
```

```python
# test data new
new_data = np.array([
    [1,1]
])

predictions = model.predict(new_data,verbose=0)
print(predictions)

if predictions > 0.5:
    print('True')
else:
    print('False')
```
