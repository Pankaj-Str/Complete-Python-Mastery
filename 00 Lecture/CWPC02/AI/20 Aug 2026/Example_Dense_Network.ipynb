```python
# dataset -
# study hours
# attendance
# output - 1 pass
# output - 2 fail

import numpy as np

X = np.array([
    [1,50],
    [2,55],
    [3,60],
    [4,65],
    [5,70],
    [6,75],
    [7,80],
    [8,85],
]) # input

y = np.array([
    [0],
    [0],
    [0],
    [1],
    [1],
    [1],
    [1],
    [1],
]) # output
```

```python
# import Dense Network
from tensorflow.keras.layers import Dense
# import Sequential model
from tensorflow.keras.models import Sequential
```

```python
# create a model
model = Sequential([
    Dense(4, activation='relu', input_shape=(2,)),
    Dense(1, activation='sigmoid') # output 0 / 1
])
```

```python
# model compile
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
model.fit(X,y,epochs=100)
```
