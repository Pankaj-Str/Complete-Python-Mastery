```python
# install tensorflow

import tensorflow as tf
from tensorflow.keras import  Sequential
from tensorflow.keras.layers import Dense , Dropout
from tensorflow.keras.regularizers import l2
```

```python
# create a neural network
model = Sequential([
    # first layer + L2 Regrlarization
    Dense(
        64,
        activation='relu',
        kernel_regularizer=l2(0.01),
        input_shape=(10,)
    ),

    # Dropout
    Dropout(0.5),

    # second layer + L2 Regrlarization
    Dense(
        32,
        activation='relu',
        kernel_regularizer=l2(0.01)
    ),

    # Dropout
    Dropout(0.3),

    # Output layer
    Dense(1, activation='sigmoid')
])
```

```python
# Compile model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# show model structure
model.summary()
```

```python
# Simple Traning
import numpy as np

# example data
X = np.random.rand(1000,10)

y = np.random.randint(0,2 , size=(1000,))

# train model
history = model.fit(
    X,
    y,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)
```
