```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

sns.set_theme(style='whitegrid',palette='tab10')
# 'white', 'dark', 'whitegrid', 'darkgrid', 'ticks'
# https://seaborn.pydata.org/tutorial/color_palettes.html
plt.rcParams['figure.figsize'] = (10,6) # Default figure size
```

```python
# loading dataset
tips = sns.load_dataset('tips')
tips.head()
```

```python
print(tips.info())
```

```python
# Relational plot (total -> tips)
# simple scatter plot
sns.scatterplot(data=tips,x='total_bill',y='tip')
plt.title("total bill vs tips")
plt.show()
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',
                hue="sex")
plt.title("total bill vs tips")
plt.show()
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',
                
                style="smoker")
plt.title("total bill vs tips")
plt.show()
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',
                hue="sex",
                style="smoker")
plt.title("total bill vs tips")
plt.show()
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',
                size="size", # size of dots by party size
                palette="deep",
                hue="sex",
                style="smoker"
                )
plt.title("total bill vs tips")
plt.show()
```

```python
flights = sns.load_dataset("flights")
flights.head()
```

```python
# line plot
sns.lineplot(data=flights,
             x="year",
             y="passengers",
             hue="month")
plt.title("Airline Passrngers Over Year")
plt.show()
```

```python
tips.head()
```

```python
# relplot

sns.relplot(data=tips,
            x="total_bill",
            y="tip",
            hue="sex",
            col="time",
            row="day")
plt.show()
```

```python
# histogram plots

sns.histplot(data=tips,
             x="total_bill")
plt.title("Distribution of total bill")
plt.show()
```

```python
# Separate distribution by category
sns.histplot(data=tips,
             x="total_bill",
             hue="sex",
             bins=70)
plt.title("Distribution of total bill")
plt.show()
```

```python
# bar plot
sns.barplot(data=tips,
            x="day",
            y="total_bill",
            hue="sex")
plt.title("Average total bill by Day and sex")
plt.show()
```
