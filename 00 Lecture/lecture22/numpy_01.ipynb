```python
# import numpy as np
from numpy import random
x = random.randint(100)
print(x)
```

```python
# generate Random Float
x = random.rand()
print(x*89)
```

```python
# random integer array
x = random.randint(1000,size=(10))
print(x)
```

```python
# 2d array (row,col)
x = random.randint(100,size=(5,4))
print(x)
```

```python
# float array 
x = random.rand(5)
print(x)
# 2d array 
x = random.rand(3,5)
print(x)
```

```python
# choice item 
x = random.choice([21,22,23,24,25,26,27])
print(x)

# 2d array choice item
x = random.choice([21,22,23,34,35,36,16,17],size=(3,5))
print(x)
```
