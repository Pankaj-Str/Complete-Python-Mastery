```python
# summary 
print(df.describe())
```

```python
# find the missing values 
print(df.isnull().sum())
```

```python
df['Merchant'] = df['Merchant'].fillna('Unknown')
print(df.isnull().sum())
```

```python
df['Amount'] = df['Amount'].fillna(df['Amount'].median())
print(df.isnull().sum())
```
