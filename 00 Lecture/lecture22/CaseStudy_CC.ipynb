```python
import pandas as pd

df = pd.read_csv("credit_card_transactions.csv")
df.head() # show only 5 row
```

```python
# find the size 
print(df.shape) # (rows,columns)

# print columns name list
print(df.columns.to_list())

# find the all into 
print(df.info())
```

```python
# summary 
print(df.describe())
```

```python
# find the missing values 
print(df.isnull().sum())
```

```python
df['Merchant'] = df['Merchant'].fillna('Unknown')
df['Transaction_Type'] = df['Transaction_Type'].fillna('Unknown')
df['Category'] = df['Category'].fillna('Unknown')
df['Payment_Mode'] = df['Payment_Mode'].fillna('Unknown')
df['Transaction_Status'] = df['Transaction_Status'].fillna('Unknown')
df['Location'] = df['Location'].fillna('Unknown')
print(df.isnull().sum())
```

```python
df['Amount'] = df['Amount'].fillna(df['Amount'].median())
print(df.isnull().sum())
```

```python
df = df.dropna(subset='Customer_ID')
df = df.dropna(subset='Transaction_Date')
print(df.isnull().sum())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
# convert Date in Datetime Datatype
df["Transaction_Date"] = pd.to_datetime(df["Transaction_Date"],errors="coerce")
print(df.dtypes)
```

```python
df["Year"] = df["Transaction_Date"].dt.year
df["Month"] = df["Transaction_Date"].dt.month
df["Day"] = df["Transaction_Date"].dt.day
df.head()
```

```python
# transaction in JAN
jan_2023 = df[(df["Year"] == 2023) & (df["Month"] == 1)]
jan_2023.head()

jan_2023.to_csv("jan_2023.csv")
```

```python
# Find transactions where Amount > 5000 and Transaction_Type is "Online".
high_online_transactions = df[(df['Amount'] > 5000) & (df['Transaction_Type'] == "Online")]
print("total transactions = ",high_online_transactions.shape)
print(high_online_transactions.head())
high_online_transactions.to_csv("high_online_transactions.csv")
```

```python
# Data Manipulation and Feature Engineering
# Create a new column Discounted_Amount, assuming a 5% discount on all transactions above $5000.

df['Discounted_Amount'] = df['Amount'].apply(lambda x : x * 0.95 if x > 5000 else x)
df[['Amount','Discounted_Amount']].tail() # show last 5 rows
```

```python
df.head()
```

```python
# Categorize the Transaction Amount into Low, Medium, and High based on:
# Low: Below $1000
# Medium: Between $1000 - $7000
# High: Above $7000

df['Amount_Categorize'] = pd.cut(df['Amount'],
                                 bins=[-float('inf'),1000,7000,float('inf')],
                                 labels=['Low','Medium','High'])
df[['Amount','Amount_Categorize']].tail()
```

```python
# Find the total transaction amount per Category.
total_amount_per_category = df.groupby('Category')['Amount'].sum().reset_index()
print(total_amount_per_category)
```
