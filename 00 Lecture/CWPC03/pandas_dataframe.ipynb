```python
data = {
    'Name':['Amit','Neha','Rupali','Neelam','Nishant','Kiran','Rahul','Rohan'],
    'Age':[23,34,45,56,23,10,90,34],
    'City':['Mumbai','Jaipur','Surat','Goa','Pune','Bhuj','Jodhpur','Delhi'],
    'Salary':[2000,34000,45000,78000,4500,76000,56000,45900],
    'Department':['HR','Sales','IT','Finance','HR','Sales','IT','Finance']
}

df = pd.DataFrame(data)
# print(df)
df
```

```python
# From List of Lists 

df2 = pd.DataFrame(
    data =[['Nishant',25],['Joy',30]],
    columns =['Name','Age'],
    index = ['emp1','emp2']
)

print(df2)
```

```python
# From List of Dictionaries

records = [
    {'Name':'Joy','Age':25,'City':'Mumbai'},
    {'Name':'Nishant','Age':30,'City':'Pune'}
]

df3 = pd.DataFrame(records)
print(df3)
```

```python
# Empty DataFrame 

df_empty = pd.DataFrame(columns=['Name','Age','Salary'])
print(df_empty)
```
