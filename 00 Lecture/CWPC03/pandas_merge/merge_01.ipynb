```python
import pandas as pd
```

```python
customers = pd.read_csv("DataSet/customers_info.csv")
customers.head()
```

```python
print(customers.shape)
```

```python
orders = pd.read_csv("DataSet/orders_info.csv")
orders.head()
```

```python
print(orders.shape)
```

```python
# pd.merge(
#     left,
#     right,
#     on = "column_name",
#     how = "inner"
# )
```

```python
# inner join 
result = pd.merge(
    customers,
    orders,
    on="customer_id",
    how = "inner"
)
result.head()
```

```python
print(result.shape)
```

```python
# left join 
left_result = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="left"
)
left_result.tail()
```

```python
print(left_result.shape)
```

```python
no_orders = left_result[
    left_result["order_id"].isna()
]
#print(no_orders.head())
print(no_orders.shape)
```

```python
# Right join
right_result = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="right"
)
# right_result.head()
print(right_result.shape)
```

```python
# outer join 
outer_result = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="outer"
)
# print(outer_result.head())
print(outer_result.shape)
```

```python
# report 
result = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="inner"
)

report = result[
    [
    "customer_id",
    "full_name",
    "email",
    "order_id",
    "product_name",
    "quantity",
    "total_amount_inr",
    "order_status"
    ]
]
```

```python
report.head()
```

```python
# calculate total customer sales
customer_sales = (
    result.groupby(
        ["customer_id","full_name"]
    )["total_amount_inr"].sum().reset_index()
)

print(customer_sales.head())
```

```python
customer_sales= customer_sales.sort_values(
    by="total_amount_inr",
    ascending=False
)

customer_sales.tail(10)
```
