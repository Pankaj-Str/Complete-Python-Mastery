```python
# %pip install pandas

import pandas as pd

data = {
    "Student_id":["S101","S102","S103","S104","S105","S106"],
    "Name":["Rahul","Nishant","Riya","Amit","Payal","Neelam"],
    "Course":["Python","Java","SQL","GO","C++","JSP"],
    "City":["Delhi","Mumbai","Pune","Delhi","Jaipur","Surat"],
    "Marks":[45,34,90,56,78,77]
}

students = pd.DataFrame(data)
students
```

```python
#check index 
print(students.index)
```

```python
# access data from index 
# loc[indexnumber]
print(students.loc[1])
```

```python
# set index - Student_id
# custom index
students = students.set_index("Student_id")
students
```

```python
# access data row from custom index 
# loc[]
print(students.loc["S102"])
```

```python
# Multiple index labels

print(students.loc[["S102","S106"]])
```

```python
# series index 

marks = pd.Series(
    [85,92,78],
    index=["S101","S102","S103"]
)
marks
```

```python
# access series index 
print(marks.index)
```

```python
print(marks.index[0])
```

```python
marks.index[0] = "S999"
```

```python
# Replace complete index
marks.index = ["A","B","C"]
print(marks)
```

```python
# duplicate index 
students_dup = pd.DataFrame(
    {
        "Name":["Rahul","Nishant","Neelam","Amit"],
        "Marks":[67,89,66,50]
    },
    index=["S101","S102","S102","S103"]
)

students_dup
```

```python
# select duplicate index 
print(students_dup.loc["S102"])
```

```python
# is_unique

print(students_dup.index.is_unique)
```

```python
print(students.index.is_unique)
```

```python
# is_monotonic_increasing -
# is_monotonic_decreasing - 

s = pd.Series([10,8,7,6,5,5])
# print(s)
print(s.is_monotonic_increasing)
print(s.is_monotonic_decreasing)
```
